The purpose of this assignment was to upload to GitHub a Claude integrated jupyternotebook environment and create a query function that takes questions as string inputs,feeds it to Claude and outputs a response within the jupyternotebook. The Claude model integrated into the Jupyter notebook for all queries was the claude-haiku-4-5-20251001".
Three physics problems from quantum mechanics and astrophysics textbooks[1][2]were placed as inputs in the query function[3][4], and the outputted numerical result has been featured in the document and compared with the true calculated results from their respective textbooks. The discrepancy percentage was calculated using this formula below. 
percent_error = (abs(C - T) / T)* 100
C = Claude Value B = Textbook Value
The contents of this document are the results of rerunning the jupyternotebook script four separate times to ensure the accuracy/consistency of the statements made in this document.


```python
import os
from dotenv import load_dotenv
from anthropic import Anthropic
load_dotenv()
```

```python
client = Anthropic()  # automatically reads ANTHROPIC_API_KEY from environment
```

```python
def query_llm(prompt: str, model: str = 'claude-haiku-4-5-20251001'):
    """ This function takes a prompt input typed in by the user as a string and feeds the prompt to the Claude Haiku 4.5 LLM and returns a text response as a result"""
    client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))
    if not os.environ.get("ANTHROPIC_API_KEY"):
        print("Set ANTHROPIC_API_KEY to run this cell.")
    else:
        message = client.messages.create(
            model= model,
            max_tokens=2048,
            messages=[
                {"role": "user",
                 "content": prompt,
                }
            ],
        )
    for block in message.content:
        if block.type == "text":
            print(block.text)
```

```python
question_1 = """When two ultraviolet beams of wavelength ( lambda_1 = 80nm and lambda_2 = 110nm) 
fall on a lead surface they produce photoelectrons with maximum energies of 11.390eV and 7.154eV respectfully. 
Estimate the numerical value of Planck's constant given these values.Calculate the cutoff frequency and wavelength of lead."""
#Page 22 of Quantum Mechanics  Concepts and Applications 3rd Edition Nourdine zettili Wiley 
response_physics_question_1 = query_llm(question_1)
```

### ACCEPTED TEXTBOOK VALUE AND DISCREPENCY

The percent discrepancy values for Planck’s constant, Work Function Cutoff Frequency and Cutoff Wavelength are small enough for the Claude output to be considered correct.



```python
question_2 = """Assuming for simplicity that the Moon's orbit is exactly circular, 
we can calculate the centripetal acceleration of the Moon rapidly.
In the case r is the distance from the center of Earth to the center of the moon,
r = 3.84401x10^8m, and v is the Moon's orbital velocity is given by v. P is the period 27.3 days
find the acceleration of the moon."""
# page 36 Introduction to Modern Astrophysics 2nd Edition Bradley W. Carroll and Dale A. Ostlie 
response_physics_question_2 = query_llm(question_2)
```

### ACCEPTED TEXTBOOK VALUE AND DISCREPENCY
The Introduction to Astrophysics textbook value used for the moon's centripetal acceleration [2] calculated based on the given parameters is 0.0027 ms-2 resulting in a 1.1% percent discrepancy between the true textbook value and the Claude response value of 0.00273 ms-2; therefore, this Claude response is considered correct and the discrepancy can be attributed to how the textbook decided to round the values.


```python
question_3a = "what is the maximum angular momentum of a solar-mass black hole in SI units?"
# page 640 stars textbook example 17.3.2 Introduction to Modern Astrophysics 2nd Edition Bradley W. Carroll and Dale A. Ostlie 
response_physics_question_3a = query_llm(question_3a)
```

### ACCEPTED TEXTBOOK VALUE AND DISCREPENCY
The true value given by the introduction to astrophysics textbook and by inputting the parameters into the formula is 8.81E41 kgm^2s^-1. This is a direct result of how Claude uses an attention method to calculate/ output its next result, as each key/query was being produced/generated there must have been an issue with how the exponents were multiplied together resulting in orders of magnitude discrepencies between the true value and the claude ouput. The probability function towards the end of this key/query matrix calculations also must have been stretched very thin as the output varies between 1E48 and 1E50 much larger than the true values order of magnitude 1e41.
